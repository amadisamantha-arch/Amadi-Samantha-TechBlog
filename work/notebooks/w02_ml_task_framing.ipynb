## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

# My lane is Refresh / Content Opportunity Scoring. As an ML task, this is a
# scoring/ranking problem built on top of binary classification. I'm not just
# labeling pages as "decline" or "not decline" — I'm producing a ranked list
# ordered by likelihood of decline, so a reviewer can work down the list from
# most to least urgent. Under the hood, this uses a classifier's predicted
# probability as the ranking score, evaluated with ranking metrics
# (Precision@K) rather than plain accuracy.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

# Target/proxy: is_declining_label = (trend_direction == "down")
#
# This is a proxy, not a true future outcome — it's a bucket calculated from
# the current window, not something that happened after a defined decision
# point. A stronger version for the capstone would use a future-looking label,
# like: features from the prior 90 days -> decline over the next 30 days.
# For this Week 2 framing exercise, I'm using the starter proxy label since
# it's what the starter dataset ships with.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

# Success metric: Precision@50 (and Precision@20 as a secondary check).
#
# This matches how the output is actually used: a reviewer has limited
# capacity and will only look at a fixed number of top-ranked pages. Precision@50
# asks "of the top 50 pages I'd tell a reviewer to check first, how many are
# actually declining?" This is more meaningful here than overall accuracy,
# which would be dominated by the many pages nobody needs to review.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
# One row = one page (content item).
lane_slice = df[["content_id", "impressions_90d", "days_since_last_update",
                  "trend_direction", "avg_position", "word_count"]].copy()
lane_slice["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

print(f"Unit of analysis: one row = one page. Total pages: {len(lane_slice)}")
lane_slice.head(10)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# Evidence from the Week 1/2 starter notebooks:
print("Hand-written rule Precision@50: 0.240 (~12 of top 50 correct)")
print("Random forest Precision@50: 0.740 (~37 of top 50 correct)")
print("The learned model found combinations of signals (impressions, content age,")
print("position, freshness) that a single hand-written rule couldn't capture —")
print("roughly a 3x improvement in how many of the top 50 flagged pages were")
print("actually declining.")
```

# A fixed rule can only combine a small number of conditions a human can hold
# in their head at once. With dozens of signals interacting (impressions,
# staleness, position, word count, CTR), a model can find patterns a human
# author wouldn't think to hand-code — and the ~3x lift on Precision@50 backs
# this up directly on this data, not just in theory.
